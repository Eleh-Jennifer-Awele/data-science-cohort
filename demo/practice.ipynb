Python data types define the kind of value a variable can store and what operations can be performed on it.


```python
#1. string: this are textual data(str)
#example of string data
school = "Royal Regency College"
subject = "Government"
print(school)
print(subject)
```

```python
#2.Numerical: Stores whole numbers.(integers,int) 
Age = 87
print(type(Age))
print(Age)
```

```python
#3.floating point: Stores decimal numbers(float)
score = 76.8
school_fees = 5000.50
print(type(score))
print(school_fees)
```

```python
#4. boolean: Stores True or False values(bool)
is_student = True
is_teacher = False
print(type(is_student))
print(is_student)
```

```python
#5. list: Stores multiple items in a single variable(list). it uses square brackets []
courses =["python", "java", "javascript", "html", "css"]
print(courses)
print(type(courses))
```

```python
#6. tuple: A tuple also stores multiple items in a single variable, but unlike a list,
# you cannot change its contents after creating it.
#Tuples use parentheses ()
food = ("rice", "beans", "yam", "plantain", "egusi", "fufu")
print(food)
print(type(food))
```

```python
#7. A dictionary stores information in key-value pairs.
# It uses curly brackets {}
student = {
    "name": "Eleh Jennifer Awele",
    "school": "Ts Academy",
    "courses": "Data Science",
    "year": 2026
}
print(student)
print(student["name"])
print(type(student))
```

```python
#8. A set stores unique values. It automatically removes duplicates.
score = {80,303,50,50,80,90,100,100,90,20,20,30,40,50}
print(score)
print(type(score))
```
